### $\omega$-$\kappa$ spectrum

We proceed our analysis using Norkyst by incorporating now another spectral analysis. Previously, in the Lagrangian experiments, we evaluated the trajecotry simulations using the rotary spectra analysis and we saw how the counterclockwise (CCW) and clockwise (CW) components vary for the different regions and models.

Even though the rotary spectra provides information about the energy density along the different frequencies ($\omega$), nothing can be inferred with regards its spatial distribution. For this reason, one must evaluate the model outputs in the wavenumber dimension ($\kappa$). 

---

The main idea is as follows:

1) Consider you have a three-dimensional variable $\phi$(x,y,t), which in this case are both velocity components *u (x,y,t)* and *v (x,y,t)*.
2) For each subregion redined in the Norkyst domain, we select a grid point (X,Y) which represents the center of a tile with dimensions NxN, where N is the number of grid points in the tile (say 100). We create many of such tiles within each subregion, with a 80% overlap between them, avoiding overlapping with the land.
3) For each tile, we employ a three-dimensional Fourier transform in the (x,y,t) domain, which provides information about the zonal wavenumber (k$_{x}$), meridional wave number (k$_{y}$), and frequency ($\omega$).
4) The resulting three-dimensional spectrum was then converted into a two-dimensional spectrum by combining the zonal and meridional wave numbers into an isotropic wave number:

$$
k = \sqrt{k_x^2 + k_y^2}
$$

5) In simple terms, an evenly spaced set of isotropic wave numbers was defined, and the three-dimensional Fourier spectrum was then averaged azimuthally over all wave-number directions. This averaging was repeated for each frequency, $\omega$, producing the final two-dimensional $\omega$–$\kappa$ spectrum. 

#### Plot domain tiles

We will build and plot all the tiles for each subregion defined within the Norkyst domain

```python
import xarray as xr
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import matplotlib.colors as mcolors
import matplotlib.cm as cm
import numpy as np
import glob
import cmocean as cm
import re
from matplotlib.colors import LogNorm

def build_boxes_by_region(df, Df, varname, dx, dy=None, overlap=0.5):

    if dy is None:
        dy = dx

    if not (0 <= overlap < 1):
        raise ValueError("overlap must be between 0 and <1")

    da = df[varname][0, :]
    region = Df["region"]

    ny, nx = da.shape
    box_h = 2 * dy
    box_w = 2 * dx

    step_y = max(1, int(box_h * (1 - overlap)))
    step_x = max(1, int(box_w * (1 - overlap)))

    boxes_by_region = {}

    for cy in range(dy, ny - dy, step_y):
        for cx in range(dx, nx - dx, step_x):

            y0 = cy - dy
            y1 = cy + dy - 1
            x0 = cx - dx
            x1 = cx + dx - 1

            if y0 < 0 or x0 < 0 or y1 >= ny or x1 >= nx:
                continue

            sub = da.isel(
                {da.dims[0]: slice(y0, y1 + 1),
                 da.dims[1]: slice(x0, x1 + 1)}
            )

            if np.isnan(sub).any():
                continue

            reg_id = region[cy, cx].item()
            if np.isnan(reg_id):
                continue

            reg_id = int(reg_id)

            box_info = {
                "center_index": (cy, cx),

                "ll": (
                    df.lon[y0, x0].item(),
                    df.lat[y0, x0].item()
                ),
                "ul": (
                    df.lon[y0, x1].item(),
                    df.lat[y0, x1].item()
                ),
                "ur": (
                    df.lon[y1, x1].item(),
                    df.lat[y1, x1].item()
                ),
                "lr": (
                    df.lon[y1, x0].item(),
                    df.lat[y1, x0].item()
                ),

                # optional: store bounds too
                "bounds_index": (y0, y1, x0, x1),
            }

            if reg_id not in boxes_by_region:
                boxes_by_region[reg_id] = []

            boxes_by_region[reg_id].append(box_info)

    return boxes_by_region

# ----------------------- #
# --- Load Norkyst V3 --- #
# ----------------------- #

path_nk = 'https://thredds.met.no/thredds/dodsC/fou-hi/norkystv3_800m_m00_be'
df = xr.open_dataset(path_nk, engine='netcdf4').sel(time=slice('2026-07-31', '2026-08-01')).isel(depth=0)

# ----------------------------- #
# --- Define and merge ROIs --- #
# ----------------------------- #
path = './datasets/roi/'
file = 'roi.nc'

Df = xr.open_dataset(path+file, engine='netcdf4')

# ------------------------------- #
# --- Define boxes per region --- #
# ------------------------------- #

boxes_per_region = build_boxes_by_region(
    df,
    Df,
    varname="u_eastward",
    dx=100, # Number of grid points from the center of each tile, i.e., C-Nx to C+Nx, and C-Ny to C+Ny  
    overlap=0.8 # Define overlap percentage
)

# ----------------------------------------------- #
# --- Define here the boundaries of each tile --- #
# ----------------------------------------------- #

patches = []
region_values = []

for reg_id, boxes in boxes_per_region.items():

    for b in boxes:

        lon_poly = [
            b["ll"][0],
            b["ul"][0],
            b["ur"][0],
            b["lr"][0],
            b["ll"][0]
        ]

        lat_poly = [
            b["ll"][1],
            b["ul"][1],
            b["ur"][1],
            b["lr"][1],
            b["ll"][1]
        ]

        poly = mpatches.Polygon(list(zip(lon_poly, lat_poly)), closed=True)
        patches.append(poly)
        region_values.append(reg_id)
```

#### Calculate $\omega$-$\kappa$ spectrum

Having defined our tiles, we will exemplify the application using a single tile for a 2-month period. We will show later how each different differs from each other, and how the spectrum look like in Summer and Winter. \
For that, we will use the *xrft* library. You can read its further documentation about [here](https://github.com/xgcm/xrft). Aditionally, in order to crop the model outputs into the selected tile domain, we use *pyresample*.\
This process can take many minutes depending on the size of your tile and your time series span.

```python
import pyresample
import xrft

'''
From Spencer-Jones:
https://github.com/cspencerjones/separating-balanced/blob/main/python/plotting_scripts/spectral_plots.ipynb
'''
# Add the isotropic calculation (k,l) -> (k_r) (azimuthally average)
def azimuthal_avg(k, l, f, N, nfactor):
    """
    Takes the azimuthal average of a given field.
    """
    k = k.values; l = l.values
    kk, ll = np.meshgrid(k, l)
    K = np.sqrt(kk**2 + ll**2)
    nbins = int(N/nfactor)
    if k.max() > l.max():
        ki = np.linspace(0., l.max(), nbins)
    else:
        ki = np.linspace(0., k.max(), nbins)

    kidx = np.digitize(np.ravel(K), ki)
    area = np.bincount(kidx)

    kr = np.bincount(kidx, weights=K.ravel()) / area

    if f.ndim == 2:
        iso_f = np.ma.masked_invalid(np.bincount(kidx,
                                    weights=f.data.ravel())
                                    / area) * kr
    else:
        raise ValueError('The data has too many or few dimensions. '
                        'The input should only have the two dimensions '
                        'to take the azimuthal averaging over.')

    return kr, iso_f

def iso_spec(ds):
    
    #Zlen = len(ds.Z)
    klen = len(ds.freq_X)
    omegalen = len(ds.freq_time)
    
    nfactor = 4. 
    ds_iso = np.zeros([omegalen ,  int(klen/nfactor+1)])

    for j in range(0,omegalen):
            kr, ds_iso[j,:] = azimuthal_avg(ds.freq_X, ds.freq_Y, ds.isel(freq_time=j), 
                            len(ds.freq_X), nfactor)
    kr[0] = 0.
    ds_iso_xr = xr.DataArray(ds_iso, coords={'freq_time':ds.freq_time.values, 'kr':kr }, 
                             dims=['freq_time','kr'])
    
    return ds_iso_xr

# Open an longer series of Norkyst
df = xr.open_dataset(path_nk, engine='netcdf4').sel(time=slice('2026-08-01', '2025-08-31')).isel(depth=0)

# ------------------------------------------------- #
# --- Define the tile for an specific subdomain --- #
# ------------------------------------------------- #

REGION = 2
i = 10
reg = boxes_per_region[REGION]

# - Crop domain given the corners - # 
llcrnrlon = reg[i]['ll'][0]; llcrnrlat = reg[i]['ll'][1]; 
lrcrnrlon = reg[i]['lr'][0]; lrcrnrlat = reg[i]['lr'][1]
ulcrnrlon = reg[i]['ul'][0]; ulcrnrlat = reg[i]['ul'][1]
urcrnrlon = reg[i]['ur'][0]; urcrnrlat = reg[i]['ur'][1]


ds_geo = pyresample.geometry.GridDefinition(lons=df.lon, lats=df.lat)
pos_geo = pyresample.geometry.SwathDefinition(lons=[llcrnrlon, lrcrnrlon, urcrnrlon, ulcrnrlon], lats=[llcrnrlat, lrcrnrlat, urcrnrlat, ulcrnrlat])

_, valid_output_index, index_array, distance_array = \
                        pyresample.kd_tree.get_neighbour_info(
                            source_geo_def=ds_geo,
                            target_geo_def=pos_geo,
                            radius_of_influence=800,
                            neighbours=1)

index_array_2d = np.unravel_index(index_array, ds_geo.shape)
(eta_indices, xi_indices) = index_array_2d[0], index_array_2d[1]

eta_min, eta_max = eta_indices.min(), eta_indices.max()  # 256–603
xi_min, xi_max   = xi_indices.min(), xi_indices.max()    # 19–368

cropped_ds = df.isel(
    Y=slice(eta_min, eta_max + 1),
    X=slice(xi_min, xi_max + 1)
)

# --- We estimate the spectrum here --- #
uo = cropped_ds.u_eastward.drop_vars(['lat', 'lon', 'depth'])
vo = cropped_ds.v_northward.drop_vars(['lat', 'lon', 'depth'])

u2_psa = xrft.power_spectrum(uo, dim=['X','Y','time'],window=True,window_correction=True).compute()
v2_psa = xrft.power_spectrum(vo, dim=['X','Y','time'],window=True,window_correction=True).compute()

vel_psa = (u2_psa + v2_psa)*0.5
ps_isoa = iso_spec(vel_psa)

# --- Define key variables to be stored --- #
kr = ps_isoa.kr # Wavenumber
omega = ps_isoa.freq_time*3600 # Frequencies
mag = (np.abs(omega)*kr*ps_isoa).values # Magnitude
```

#### Plotting

Once all the calculations are done, we can finally plot the results

```python
# Define some plotting options here
lev_exp = np.linspace(-5, -1, 30)
lev = np.power(10, lev_exp)[0:30]
norm = mcolors.LogNorm(vmin=lev.min(), vmax=lev.max())


# Plot here
fig, ax = plt.subplots(1,1, dpi=200)

cs = ax.contourf(kr*1e3, omega, mag, lev, norm=LogNorm(), cmap=cm.cm.deep, extend='both')

# Adjust scales
ax.set_xscale('log')
ax.set_yscale('log')
ax.set_xlim(0.019, 0.6)
ax.set_ylim(0.001, 0.5)

# Set general y-label
fig.supylabel('Frequency [cph]', size=12)

# Set x-labels
ax.set_xlabel('Horizontal \n wavenumber [cpkm]')
ax.set_xlabel('Horizontal \n wavenumber [cpkm]')

# Plot tides etc
cph = 0.000277777777777778
s1 = (1/(24*3600))/cph
m2 = (1/(12.42*3600))/cph
m4 = (1/(6.2103*3600))/cph

# Define color for texts and dashed lines
Color = '#822C0D' #'#AA353F'
xcoords = [s1, m2, m4]


for xc in xcoords:
    ax.axhline(y=xc, c=Color, ls='--', lw=0.6, color='r', alpha=1, zorder=10)


# Inertial band
f_values = np.arange(50, 85, 0.5)
OMEGA = 7.2921*1e-5
f = 2*OMEGA*np.sin(np.deg2rad(f_values))
T = (2*np.pi)/f

# Plot the region
f0 = (1/(T[0]/3600))
f1 = (1/(T[-1]/3600))

ax.axhspan(f0, f1, alpha=.2, facecolor='gray', edgecolor='gray')

fig.subplots_adjust(right=0.8)
cbar_ax = fig.add_axes([0.82, 0.24, 0.02, 0.5])
cbar = fig.colorbar(cs, ticks=[10**-5, 10**-4, 10**-3, 10**-2, 10**-1, 10**0, 10**1], cax=cbar_ax)
cbar.ax.tick_params(labelsize=12) # Set the desired font size
cbar.set_label(r'$K \times \omega \times |\hat{E}_{\mathrm{EKE}}|^{2}$', size=12)
```

#### Interpration

What information can we obtain from this plot? Since it displays how energy is distributed through different wavenumbers and frequencies, we can evaluate which type of phenomena is predominant in the area of interest.\
We can see that hte higher energy regions (blue) are located at slow (low frequency) and larger (small wavenumbers) bands, with a notable peak around the inertial frequency (gray brand) and the semi-diurnal tidal band (mid red line).

```python
# Define location of the calculated variables
path_to = './datasets/omega_kappa/'
files = glob.glob(path_to+'*.npz')


###################### 
### Make plot here ###
######################

# Define layout
regions = range(2, 7)           # rows (2..6)
seasons = ['jfm', 'aso']        # columns

# Define some extra stuff for colorbar
lev_exp = np.linspace(-5, 0, 30)
lev = np.power(10, lev_exp)[0:30]
norm = mcolors.LogNorm(vmin=lev.min(), vmax=lev.max())

# We define spine color according to ROI #
cmap = plt.get_cmap("Dark2", 7)
bounds = np.arange(1, 7 + 1, 1)
Norm = mcolors.BoundaryNorm(bounds, cmap.N)

# Init figure 
fig, axes = plt.subplots(5, 2, figsize=(9,10), sharey=True, sharex=True)

i = 2
for f in files:
    m = re.search(r"region(\d+)_(\w+)_", f)
    region = int(m.group(1))
    season = m.group(2)

    row = regions.index(region)
    col = seasons.index(season)

    ax = axes[row, col]

    color = cmap(Norm(row + 2))   # color for that row

    for spine in ax.spines.values():
        spine.set_edgecolor(color)
        spine.set_linewidth(2)
    
    df = np.load(f)
    
    # Define variables here
    kappa = np.nanmean(df['kappa'], axis=0)
    omega = np.nanmean(df['omega'], axis=0)
    spec = np.nanmean(df['spec'], axis=0)
    
    cs = ax.contourf(kappa*1e3, omega, spec, lev, norm=LogNorm(), cmap=cm.cm.deep, extend='both')

    # Adjust scales
    ax.set_xscale('log')
    ax.set_yscale('log')
    ax.set_xlim(0.01, 0.6)
    ax.set_ylim(0.001, 0.5)

# Set general y-label
fig.supylabel('Frequency [cph]', size=12)

# Set x-labels
axes[4, 0].set_xlabel('Horizontal \n wavenumber [cpkm]')
axes[4, 1].set_xlabel('Horizontal \n wavenumber [cpkm]')

# Set title to columns
axes[0,0].set_title('Jan-Feb-Mar', size=12)
axes[0,1].set_title('Aug-Sep-Oct', size=12)

# Plot tides etc
cph = 0.000277777777777778
s1 = (1/(24*3600))/cph
m2 = (1/(12.42*3600))/cph
m4 = (1/(6.2103*3600))/cph

# Define color for texts and dashed lines
Color = '#822C0D' #'#AA353F'
xcoords = [s1, m2, m4]

for ax in axes.flat:
    for xc in xcoords:
        ax.axhline(y=xc, c=Color, ls='--', lw=0.6, color='r', alpha=1, zorder=10)


# Inertial band
f_values = np.arange(50, 85, 0.5)
omega = 7.2921*1e-5
f = 2*omega*np.sin(np.deg2rad(f_values))
T = (2*np.pi)/f

# Plot the region
f0 = (1/(T[0]/3600))
f1 = (1/(T[-1]/3600))

for ax in axes.flat:
    ax.axhspan(f0, f1, alpha=.5, facecolor='gray', edgecolor='gray')

fig.subplots_adjust(right=0.8)
cbar_ax = fig.add_axes([0.82, 0.24, 0.02, 0.5])
cbar = fig.colorbar(cs, ticks=[10**-5, 10**-4, 10**-3, 10**-2, 10**-1, 10**0, 10**1], cax=cbar_ax)
cbar.ax.tick_params(labelsize=12) # Set the desired font size
cbar.set_label(r'$K \times \omega \times |\hat{E}_{\mathrm{EKE}}|^{2}$', size=12)
```

#### Interpretation


Three interesting points can be highlighted here:

1) There are large differences between the locations. For the orange boxes, covering mainly the North Sea and and part of the southwestern Norwegian coast, we see the predominance of peaks in discrete bands, associated to the different tidal components. As we move northwards, and especially at the 'corner' of Norway (purple boxes) and between the shelf-break (gray boxes) which also encompass the Lofoten region, we observe stronger energies associated to meso- to submesoscale motions (dark blue blobb from lower to higher wavenumbers).
2) Another interesting aspect is how this blobb is greater for all regions in winter time (Jan-Feb-Mar, left column) except for the North Sea region. Strong winter winds and heat fluxes trigger widespread mixed-layer instabilities (MLIs). This dramatically boosts meso- and submesoscale balanced motions. This injection of energy at small spatial scales causes a flattening of the horizontal wavenumber spectrum at scales below 100 km. When comparing Winter to Summer, one can see that more energy is present at higher wavenumbers (lower-right x-axes) for frequencies lower than 10$^{-2}$ cph (order of days).
3) Something very interesting can be seen in Summer: A band of high energy growin from around 12 hours (gray rectangle) towards superinertial bands ($\omega$ > f) as the wavenumber increases (small scale motions). A possible explanation relies on a type of phenomenon called internal gravity waves. Solar heating creates a shallow, strongly stratified thermocline. This suppresses the wintertime mixed-layer instabilities. Because the submesoscale EKE drops, the background balanced spectrum plunges steeply. However, high-frequency internal tides and gravity waves are still active. Rocha, Gille, et al. (2016) demonstrated that this results from the summertime intensification of higher baroclinic modes associated with the shallow mixed layer.
